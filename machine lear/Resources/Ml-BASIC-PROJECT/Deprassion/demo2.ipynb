```python
import pandas as pd
import numpy as np
df=pd.read_csv('post natal data.csv')
df
```

```python

epds_mapping = {
    'Yes': 2,                 # Moderate symptom
    'No': 0,                  # No symptom
    'Maybe': 1,               # Mild symptom
    'Sometimes': 1,
    'Two or more days a week': 2  # Example for sleep column
}

symptom_cols = [
    'Feeling sad or Tearful',
    'Irritable towards baby & partner',
    'Trouble sleeping at night',
    'Problems concentrating or making decision',
    'Overeating or loss of appetite',
    'Feeling anxious',
    'Feeling of guilt',
    'Problems of bonding with baby'
]

for col in symptom_cols:
    df[col + '_score'] = df[col].map(epds_mapping)

df['Suicide_attempt_score'] = df['Suicide attempt'].map({'Yes': 3, 'No': 0})
score_cols = [col + '_score' for col in symptom_cols] + ['Suicide_attempt_score']
print("\nConverted numeric scores:")
print(df[score_cols].head(5))
```

```python
score_cols = [col + '_score' for col in symptom_cols] + ['Suicide_attempt_score']

df['EPDS_Total'] = df[score_cols].sum(axis=1)

df['EPDS_Total_normalized'] = df['EPDS_Total'] * (30 / 27)

df['PPD_Result'] = df['EPDS_Total_normalized'].apply(lambda x: 1 if x >= 10 else 0)

df[['EPDS_Total', 'EPDS_Total_normalized', 'PPD_Result']].head(10)
```

```python
threshold = 10

df['PPD_Result'] = df['EPDS_Total_normalized'].apply(lambda x: 1 if x >= threshold else 0)

df[['EPDS_Total', 'PPD_Result']].head(10)
```

```python
df.to_csv('PPD_dataset_with_scores.csv', index=False)
```
